```python
import matplotlib.pyplot as plt
import numpy as np

# 数据集名称
datasets = ['Hospital', 'Beers', 'Rayyan']

# F1得分数据
f1_scores_with_hints = [0.94, 0.96, 0.86]
f1_scores_without_hints = [0.57, 0.13, 0.60]

# 创建一个图和一个轴
fig, ax = plt.subplots()

# 设置条形的宽度
bar_width = 0.35

# 计算条形的位置
index = np.arange(len(datasets))

# 绘制条形图
bars1 = ax.bar(index, f1_scores_with_hints, bar_width, label='LLM with hints')
bars2 = ax.bar(index + bar_width, f1_scores_without_hints, bar_width, label='LLM w/o hints')

# 在每个条形图上显示F1-score的数字
for bar in bars1:
    yval = bar.get_height()
    ax.text(bar.get_x() + bar.get_width()/2, yval + 0.02, round(yval, 2), ha='center', va='bottom', fontsize=16)

for bar in bars2:
    yval = bar.get_height()
    ax.text(bar.get_x() + bar.get_width()/2, yval + 0.02, round(yval, 2), ha='center', va='bottom', fontsize=16)

# 添加x轴和y轴标签
ax.set_ylabel('F1-Score', fontsize=16)

# 添加x轴刻度标签
ax.set_xticks(index + bar_width / 2)
ax.set_xticklabels(datasets, fontsize=16)

# 添加y轴刻度标签
ax.tick_params(axis='y', labelsize=16)

# 添加图例
ax.legend(loc='lower left', fontsize=14.5)

# 调整y轴的上限以避免数字重叠
ax.set_ylim(0, 1.10)  # 设置y轴的上限稍微高于1

# 显示图形
plt.show()

fig.savefig('hint_revise_v3.pdf', bbox_inches='tight')
```
